```python
#question 28 : 2d slicing challenge : given
matrix = np.array([
    [5,10,15,20],
    [25,30,35,40],
    [45,50,55,60]
])
#get second row only, third column only and submatrix
print(f"Second Row only : {matrix[1,:]}")
print(f"Third Column only : {matrix[:,2]}")

subrow = [0,1]
subcol = [1,2]

submatrix = matrix[np.ix_(subrow,subcol)]
print(submatrix)
```

```python
#question 29 : Scenario Based Analysis
steps = np.array([4200, 5800, 6100, 10000, 3000, 8200, 7600])
#calculate average steps
print(f"Average Steps : {np.mean(steps)}")
#find highest steps
print(f"Highest Steps : {np.max(steps)}")
#find days with more than 6000 steps
days = np.where(steps > 6000)[0] + 1
print(days)
#Increase every value by 500
steps_in5 = steps + 500
print(steps_in5)
```

```python
#question 30 : Mini Case study (Hard)
scores = np.array([
    [70, 80, 90],
    [88, 76, 95],
    [60, 75, 85]
])
# each row = student, each column = subject
# find average score of each student
print(f"AVG Student : {np.mean(scores, axis = 1)}")
# find average score of each subject
print(f"AVG Subject : {np.mean(scores, axis = 0)}")
# Highest score in the whole matrix
print(f"Max Score : {np.max(scores)}")
# Extract all scores greater than 80
print(f"Extract : {scores[scores > 80]}")
```
