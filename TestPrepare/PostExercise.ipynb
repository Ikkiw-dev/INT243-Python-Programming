```python
#Group by company_location and employment_type. Find the number of employees (count) and the maximum salary in USD for each combination. 
df_grouped = df.groupby(['company_location', 'employment_type']).agg(employee_num=('salary_in_usd', 'count'),
                                                                     max_salary=('salary_in_usd','max'))
df_grouped.sort_values(by='max_salary', ascending = False)
#Sort the results to show the locations with the highest max salaries at the top.
```

```python
#Use np.where() to create a new Boolean column called is_highly_paid. 
#Set it to True if the salary_in_usd is above the 75th percentile of the entire dataset, and False otherwise.
# df['salary_in_usd'].describe()
df['salary_in_usd'].quantile(0.75) #got 150000 from run this code

df['is_highly_paid'] = np.where(df['salary_in_usd'] > 150000, True, False)
df.head()
```

```python
df.loc[(df['salary_in_usd'] > 150000) & (df['is_highly_paid'] == False)]
```

```python
#Use np.select() or pandas pd.cut() 
#to create a new column called salary_tier that categorized salary_in_usd into three bins:
#apply condition from the notion
condition_lst = [df['salary_in_usd'] < 50000, 
                 (df['salary_in_usd'] >= 50000) & (df['salary_in_usd'] <= 150000),
                df['salary_in_usd'] > 150000]
salary_tier = ['Low','Medium','High']

df['salary_tier'] = np.select(condition_lst, salary_tier, default = 'Unknown')
df
```

```python
#Create a new column called is_expat that contains Trye if the employee_residence does not equal the company_location, and False if they match. 
#What is the average salary of an expat vs. a local?
df['is_expat'] = df['employee_residence'] != df['company_location']
df.groupby('is_expat')['salary_in_usd'].mean()
```

```python
#Filter the data to only include companies located in the US (US), UK (GB), and Canada (CA). 
#For these three countries, calculate the percentage of roles that are fully remote 
#(remote_ratio == 100) versus not fully remote.
countries = ['US','GB','CA']

df_countries = df.loc[df['company_location'].isin(countries)]
df_countries['remote_ratio'].apply(lambda x: x == 100).value_counts(normalize = True)
```

```python
#Which 5 job_titles have the highest median salary? 
#(Ensure you only consider job titles that appear at least 5 times 
#in the dataset to avoid outliers skewing the data).
job_num = df['job_title'].value_counts() >= 5
atleast5 = job_num[job_num].index

df_sel_job = df[df['job_title'].isin(atleast5)]
df_sel_job.groupby('job_title')['salary_in_usd'].median().sort_values(ascending = False).head(5)
```
